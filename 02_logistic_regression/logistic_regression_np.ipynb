```python
df['completed_50m_ultra'].replace('no', 0.0, inplace=True)
df['completed_50m_ultra'].replace('yes', 1.0, inplace=True)
df
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def binary_cross_entropy(m, b, points):
    """Binary version of the cross entropy loss function"""
    loss = 0
    n = len(points)
    for i in range(n):
        x = df.iloc[i].miles_per_week
        y = df.iloc[i].completed_50m_ultra
        sig = sigmoid(m * x + b)
        loss += -(1/n) * (y * np.log(sig) + (1 - y) * np.log(1 - sig))
    return loss
    
def gradient_descent(m_now, b_now, points, L):
    """
    Gradient descent algorithm optimizing a 
    binary cross entropy loss function
    """
    m_grad = 0
    b_grad = 0
    n = len(points)
    for i in range(n):
        x = df.iloc[i].miles_per_week
        y = df.iloc[i].completed_50m_ultra
        m_grad += (1/n) * (sigmoid(m_now * x + b_now) - y) * x
        b_grad += (1/n) * (sigmoid(m_now * x + b_now) - y)
    m = m_now - L * m_grad
    b = b_now - L * b_grad
    return m, b
```

```python
def vectorized_binary_cross_entropy(theta, x, y):
    """Binary version of the cross entropy loss function"""
    m = y.size
    sig = sigmoid(x @ theta)
    loss = -(1/m) * (y * np.log(sig) + (1 - y) * np.log(1 - sig))
    return np.sum(loss)
    
def vectorized_gradient_descent(theta, x, y, L):
    """Gradient descent algorithm optimizing a 
    binary cross entropy loss"""
    m = y.size
    grad = (x.T @ (sigmoid(x @ theta) - y)) / m
    new_theta = theta - L * grad
    return new_theta
```

```python
def predict_proba(x, theta):
    x_b = np.c_[np.ones((x.shape[0], 1)), x]
    return sigmoid(x_b @ theta)

def predict(x, theta, threshold=0.5):
    return (predict_proba(x, theta) >= threshold).astype(int)

test = np.array([[80, 60],[10, 20]])
print(predict_proba(test, theta))
print(predict(test, theta))
```
